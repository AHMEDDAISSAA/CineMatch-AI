```python
import sys
import pandas as pd
from pathlib import Path

ROOT = Path.cwd().parent
sys.path.append(str(ROOT))
from src.split import temporal_split

ratings = pd.read_csv(ROOT / "data" / "raw" / "ratings.csv")
movies = pd.read_csv(ROOT / "data" / "raw" / "movies.csv")
```

```python
train, val, test = temporal_split(ratings)
print("Train:", train.shape, "| Validation:", val.shape, "| Test:", test.shape)
print("Utilisateurs:", train["userId"].nunique(), val["userId"].nunique(), test["userId"].nunique())
```

```python
keys = ["userId", "movieId"]
print("Chevauchements:", len(train.merge(val, on=keys)),
      len(train.merge(test, on=keys)), len(val.merge(test, on=keys)))

tmax = train.groupby("userId")["timestamp"].max()
vmin = val.groupby("userId")["timestamp"].min()
vmax = val.groupby("userId")["timestamp"].max()
tmin = test.groupby("userId")["timestamp"].min()
print("Train dépasse Validation:", (tmax > vmin).sum(), "| Validation dépasse Test:", (vmax > tmin).sum())
```
